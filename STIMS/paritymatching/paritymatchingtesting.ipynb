## Complimentary Gap 

```python
#1. introduce error
#2.split into x and z
circuit = stim.Circuit.from_file('../circuits/yerrorbias5/5_0.001_y.stim')
edge_map = edge_coordinate_map(circuit)
# print(edge_map)
x_edges, z_edges = matching(circuit)
# print("X edges:", x_edges)
# print("Z edges:", z_edges)
z_error_locations, x_error_locations = get_locations_from_edges(circuit, x_edges, z_edges)

#3.get weight of solution from pymatching
solution_parity_z = len(z_error_locations) 
solution_parity_x = len(x_error_locations) 

#4.
# success_of_sshot, gap = get_data_from_shot(<data file>)
alternate_sol_parity_z =  parity_of_solution(solution_parity_z, gap)
alternate_sol_parity_x =  parity_of_solution(solution_parity_x, gap)

#choose lowest parity
x_parities = [solution_parity_x, alternate_sol_parity_x]
z_parities = [solution_parity_z, alternate_sol_parity_z]    
combined = x_parities + z_parities
odd_sum = sum(x for x in combined if x % 2 != 0)
even_sum = sum(x for x in combined if x % 2 == 0)


final_x_parity = 0 # 0 for original, 1 for alternate
final_z_parity = 0 # 0 for original, 1 for alternate
if odd_sum < even_sum:
    if x_parities[0] % 2 == 0:
        final_x_parity = 1
    if z_parities[0] % 2 == 0:
        final_z_parity = 1
if even_sum < odd_sum:
    if x_parities[0] % 2 != 0:
        final_x_parity = 1
    if z_parities[0] % 2 != 0:
        final_z_parity = 1

#applying correction based on final parities


```
